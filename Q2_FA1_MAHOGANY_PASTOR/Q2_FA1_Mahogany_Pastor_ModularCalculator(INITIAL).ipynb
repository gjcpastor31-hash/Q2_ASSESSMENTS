```python
def add_numbers(a, b):
    return a + b

def subtract_numbers(a, b):
    return a - b

def multiply_numbers(a, b):
    return a * b

def divide_numbers(a, b):
    if b == 0:
        return "Error: Division by zero is not allowed."
    return a / b

a = float(input("Enter your first number: "))
b = float(input("Enter your second number: "))

print("1 is addition,")
print("2 is subtraction,")
print("3 is multiplication,")
print("4 is division.")

operation = int(input("Enter your operation (1-4): "))

if operation == 1:
    output = add_numbers(a, b)
elif operation == 2:
    output = subtract_numbers(a, b)
elif operation == 3:
    output = multiply_numbers(a, b)
elif operation == 4:
    output = divide_numbers(a, b)
else:
    output = "Invalid operation selected."

print(output)
```
