```python
def add_numbers(num1, num2):
  #Adds the two numbers.
    return num1 + num2

def subtract_numbers(num1, num2):
  #Subtracts the two numbers.
    return num1 - num2

def multiply_numbers(num1, num2):
  #Multiplies the two numbers.
    return num1 * num2

def divide_numbers(num1, num2):
  #Divides the two numbers, and also has an extra line of code to make sure we do not accidentally divide by zero.
    if num2 == 0:
        return "Error: Division by zero is not allowed."
    return num1 / num2

#Asks the user to input their first number and also their second number.
num1 = float(input("Enter your first number: "))
num2 = float(input("Enter your second number: "))

#I give the user 4 available operations to choose from.
print("----------------------------------------")
print("[==AVAILABLE OPERATIONS==]")
print("[1] - Addition(➕)")
print("[2] - Subtraction(➖)")
print("[3] - Multiplication(✖️)")
print("[4] - Division(➗)")
print("----------------------------------------")

operation = int(input("Enter your operation (1-4): "))


if operation == 1:
    output = add_numbers(num1, num2)
elif operation == 2:
    output = subtract_numbers(num1, num2)
elif operation == 3:
    output = multiply_numbers(num1, num2)
elif operation == 4:
    output = divide_numbers(num1, num2)
else:
    output = "❌ Invalid selection. Please choose an operation from 1 to 4."


print(output)

```


1. What functions did you create in your calculator program?

I created four functions called add_numbers, subtract_numbers, multiply_numbers, and divide_numbers. These handle addition, subtraction, multiplication, and safe division respectively.


2. What parameters did your functions use?

For all my functions, the parameters were num1 and num2. These paremeters act as placeholders since the user hasn't inputted a value yet.

2. What parameters did your functions use?

Every function used two parameters named num1 and num2. They act as temporary placeholder variables to receive the values that need to be calculated.

3. What arguments were passed when the functions were called?

The real values stored in the user-input variables num1 and num2 were passed as the actual arguments during each function call.

4. How did your program use the returned value?

The program captured the result sent back by the function and stored it inside the output variable. It then printed (output) on the screen.

5. Why is it better to divide the calculator program into functions instead of writing everything in one long block of code?

Using functions makes the program much more organized, easier to read, and simpler to fix. It is also very useful when you use the same equations over and over again. Instead of having to do the equations in bulk, you can just call the function or functions.
